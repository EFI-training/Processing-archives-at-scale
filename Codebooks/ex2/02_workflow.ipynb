Example answer:

```python
from pathlib import Path

ROOT_FOLDER = Path("../raw/ans_project_data") ## example: ../raw/project_data
OUTPUT_ROOT = Path("../output")
SUPPORTED_IMAGE_TYPES = {".jpg", ".jpeg", ".png"}
MODEL = "gpt-4o-mini"


### 01 Find document folders
documents = sorted(folder for folder in ROOT_FOLDER.iterdir() if folder.is_dir())
print(f"Found {len(documents)} book folders: {[folder.name for folder in documents]}")

### 02 Extract and export each image
for document in documents:
    document_folder = document / "image"
    image_paths = sorted(image_path for image_path in document_folder.iterdir()
        if image_path.is_file()
        and image_path.suffix.lower() in SUPPORTED_IMAGE_TYPES
    )
    output_folder = OUTPUT_ROOT / document.name
    output_folder.mkdir(parents=True, exist_ok=True)
    print(f"{document.name}: {len(image_paths)} image(s)")

    for index, image_path in enumerate(image_paths[:15], start=1):
        text = extract_text(image_path, model=MODEL)
        output_path = output_folder / f"{image_path.stem}.txt"
        output_path.write_text(text + "\n", encoding="utf-8")
        print(f"[{index}/{len(image_paths)}] Extracted and saved {output_path.name}")
```
